```python
### User defined functions

### Without input without return value
### With input without return value
### Without input with return value
### with input with return value

## without input & without return value

def mul():
    a = 20
    b = 40
    c = a *b
    print(c)
mul() ## Function calling
```

```python
### With input without return value

def mul(c,d):
    e = c * d
    print(e)

mul(20,30)
# mul() # mul() missing 2 required positional arguments: 'c' and 'd'
```

```python
## Without input & Return value

def mul():
    f = 20
    g = 60
    h = f * g
    return h
mul()
```

```python
# withinput with Return value

def mul(x,y):
    z = x * y
    return z
mul(2,3)
```

## Parameters 

Collecting the inputs for the functions is called parameters

Arguments : input passing to the functions.

Types of Arguments:

Positional Arguments

Default Arguments

Keyword Arugements

Variable length Arguments

Varaible length Keywords Arguments



```python
## Positional Arguments

def power_off(a,b):
    c = a * b
    print(c)
power_off(10,20)

# Now power_off changes

# power_off(2) power_off() missing 1 required positional argument: 'b'

# To over come this we can give default arguments
```

```python
## Default Arguments

# Default Arguments are also called as optional arguments

def power_off(m,n = 2):
    return m ** n
power_off(2) # 4
power_off(2,4) # 16
```

```python
# After default Aruguments does'nt have non default arguments.

def fun(a,b=1,c): # SyntaxError: parameter without a default follows parameter with a default
    print(a,b,c)
fun(2,3,4)
```

```python
# Non default argument fallows default arguments.

def fun(a,b=1,c=2):
    print(a,b,c)
fun(3)
```

```python
## I dont want to change the positional arguments then i should go for the keyword arguments
```

```python
## Keyword arguments

def power_off(s,t):
    return s**t
power_off(s = 10,t = 2)
power_off(t = 3, s = 4)
```

```python
def pizza_toppings(toppings):
    print(toppings)

pizza_toppings("cheese")
```

```python
 # pizza_toppings("cheese","Olive","onions","corn") # pizza_toppings() takes 1 positional argument but 4 were given

# To over come this use * - > indicate vaiable length arguments.

def pizza_toppings(*toppings):
    print(toppings)

pizza_toppings("cheese","Olive","onions","corn")
```

```python
## varibale length keyword arguments

def collect_student_info(*data):
    print(data)

collect_student_info(Name = "Rohit",age = 28, Gender = "male")
```

```python
## varibale length keyword arguments

def collect_student_info(**data):
    print(data)
    print(type(data))

collect_student_info(Name = "Rohit",age = 28, Gender = "male")
```
