```python
## Type Casting 

## converting one data type to another data type is known as type casting 

## int -- > float

a = 10
print(a)
print(type(a))

# To convert int --- > float

b = float(a)
print(b)
print(type(b))
```

```python
# float --- > int # Here we can loss the information of data

c = 10.99
print(c)
print(type(c))
print(id(c))

# typecasting 

d = int(c)
print(d)
print(type(d))
print(id(d))
```

```python
# float --- > complex

e = 99.534
print(e)
print(type(e))
print(id(e))

f = complex(e)
print(f)
print(type(f))
print(id(f))
```

```python
# int --- > str
g = 98
print(g)
print(type(g))
print(id(g))

h = str(g)
print(h)
print(type(h))
print(id(h))
```

```python
# Data types are divided into two types 
# Mutable and Immutable data types

# Imutable : int,float, str, complex, bool
# Mutable : list, dict, set

i = 10
print(i)
print(type(i))
print(id(i))

j = 10

print(j)
print(type(j))
print(id(j))

print(i is j )
```

```python
## Immutable 

lst = [23,24,25,26,27]
print(lst)
print(type(lst))
lst.append(50000)
print(lst)

lst1 = [1,2,3,4,5,6,7]
print(lst1)
print(type(lst1))
print(id(lst1))
lst1.append(10000)
print(lst1)
```
