```python
# How do you access individual characters in a string

# u can access individual characters in a string using indexing. indexing allows you to acces elements in the string based on their posistion, starting with zero for the first character


# accessing individual characters in a string
```

```python
my_string = "Hello world!"

# Accessing the first character index(0)

first_char = my_string[0]

print(first_char)

# accessing the fifth charcter (index(4))

fifth_char = my_string[4]

print(fifth_char)

# Accessing the last character using negative indexing

last_char = my_string[-1]
print(last_char)
```

```python
s = input("enter a string:")

low_count,up_count,sp_count,num_count = 0,0,0,0

for i in s:
    if i.islower():
        low_count = low_count + 1
    if i.upper():
        up_count = up_count + 1
    if i.isnumeric():
        num_count = num_count + 1
    else:
        sp_count = sp_count + 1

print("Lower case=",low_count)
print("upper case",up_count)
print("Numeric=",num_count)
print("Special count=",sp_count)


    
```
