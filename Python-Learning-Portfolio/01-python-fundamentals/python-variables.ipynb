```python
### Declaring and Assigning values

age = 32
height = 6.4
Name = "Krish"
Is_student = True

print("age:",age)
print("height:",height)
print("Name:",Name)
print("Is_student:",Is_student)
```

```python
## Naming conventions 
## variable name should be descriptive
## They must start with the letter or an '_' and contains letters, numbers and underscores
## variable names are case sensitive

first_name = "Krish"
last1name = "Naik"
```

```python
## Invalid naming conventions

## @madhu = "krish"
## 2age = 32
## first-name = "Krish"
```

```python
### Understanding Variable types

## python is dyinamically typed, type of variable is determined by runtime

age = 25 # int
height = 6.2 #float
name = "Madhu" #string
Is_student = True # bool

print(type(age))
print(type(height))
print(type(name))
print(type(Is_student))
```
