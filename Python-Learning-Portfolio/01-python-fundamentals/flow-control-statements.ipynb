## Lambda

When ever lyou want to create a fucntion we have 2 key words 

def

lambda




```python
def power_off(num,p):
    return num**p
res = power_off(5,2)
print(res)

res1 = power_off(4,2)
print(res1)
```

```python
## Lambda doesnt have any name becasue of that its called ananomys function

# Syntax lambda arguments: Expression

res = (lambda num,p:num**p)(5,2)
print(res)
```

```python
## If you want to use multiple times

fun = lambda num,p:num**p
res = fun(6,2)
print(res)
res1 = fun(9,2)
print(res1)
```

General use of lambda:

filter()

filter(function,sequence)



```python
lst = [10,20,40,50,60,70,80]
def fun(x):
    if x%2 == 0:
        return True
    else:
        return False

even_list = list(filter(fun,lst))
print(even_list)
```

```python
lst = [2,3,4,5,6,7,8,9]
def fun(x):
    if x%2 == 0:
        return True
    else:
        return False

even_list = list(filter(fun,lst))
print(even_list)
```

```python
lst = [2,3,4,5,6,7,8,9]
def fun(x):
    if x%2 != 0:
        return True
    else:
        return False

odd_list = list(filter(fun,lst))
print(odd_list)
```

```python
# Reduce

from functools import reduce

lst3 = [1,2,3,4,5]
def fun(x,y):
    return x+y
res = reduce(fun,lst3)
print(res)
```
