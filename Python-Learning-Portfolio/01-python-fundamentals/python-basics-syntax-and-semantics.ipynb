#### Defination of Syntax :

Syntax is a set of rules that defines the combination of symbols that are considered to be correctly structed programs in a language. 
Arragement of words and symbols in a code


Semantics refers to the meaning or the interpretation of the symbols, characters, and commands in langugage.


```python
### Basic syntax rules in python
### Case sensitiviy -- Python is case sensitive 

name = "Krish"

Name = "KRISH"

print(name)
print(Name)

print(id(name))
print(id(Name))
```

### Identation

Python uses indentation to determine the grouping of code. Usually we use 4 spaces 

```python
## Indentation

## python uses identation to define blocks of code. we use 4 spaces or tab 

age = 32
if age>30:
    print(age)
print(age)
```
