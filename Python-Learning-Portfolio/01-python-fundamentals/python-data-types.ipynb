```python
## Basic Data Types
## Int

a = 10
print(a)
print(type(a)) ## Int
print(id(a)) ### 140725572750040
```

```python
## Float

b = 10.5
print(b)
print(type(b)) ## Float
print(id(b)) ## 2618469189424
```

```python
## String

c = "Hello"
print(c)
print(type(c)) ## <class 'str'>
print(id(c)) ## 2618469937392
```

```python
## Complex 

d = 4+5j
print(d)
print(type(d)) # <class 'complex'>
print(id(d)) # 2618475259824
```

```python
## Bool

e = True
print(e)
print(type(e)) # <class 'bool'>
print(id(e)) # 140725571619536
```

```python
## Advanced Data Types

# List

# List is a collection of heterogenous type of data

# List is mutable 

lst = [10,20,30,40,50]
lst1 = [10,10.24,"Hello world",5+6j,True]
print(lst1)
print(type(lst1))
print(id(lst1))
print(lst)
print(type(lst))
print(id(lst))


# Modifications can be done in list
# Append
# remove

lst.append(100)
print(lst)
lst1.append("Cherry")
print(lst1)

lst.remove(100)

print(lst)

lst1.remove('Hello world')

print(lst1)
```

```python
## Tuple

## Tuple can not be modified

Tup = (10,20,30)
print(Tup)
print(type(Tup))
print(id(Tup))

# Tup.append(10) AttributeError: 'tuple' object has no attribute 'append'
```

```python
## Set

# set is a unordered collection of data
# There is no index values then where you can access
# set is mutable in nature

s = {10,20,30,50,60,50}
print(s)
 #print(s[0]) TypeError: 'set' object is not subscriptable
print(type(s))
print(id(s))

s.add(600)
print(s)

s.remove(600)

print(s)
```

```python
## Dictionary

# Mapping the collection of data

d = {"Dennis Riche": "C",
     "James Gossling": "Java",
     "Gauido Van Rassum": "Python"
}

print(d)
print(type(d))
print(id(d))

## Add

d['Brendin Edge'] = "java Script"

print(d)

## Remove
d.pop("Brendin Edge")

```
