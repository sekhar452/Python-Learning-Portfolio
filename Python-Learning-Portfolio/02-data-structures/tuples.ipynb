```python
## tuple 

# tuple is a static array 

t = (10,20,30,40)

print(t)

print(type(t))
```

```python
t1 = (10,)

print(type(t1))
```

```python
t2 = ("hello",10,20,1+3j,True)

print(t2)

t3 = (10,20,30,10,20)

print(t3)

t4 = (10,[20,30,40],(50,60,70),{9,10,11},{1:"Hello",2:"World"})

print(t4)

# t4.append(10) # 'tuple' object has no attribute 'append' tuple is immutable type of data
```

```python
## set 

s = {10,20,30,40}

print(s)

print(type(s))

## A set don't maintain order
# A set don't allow duplicates
```

```python
# set don't allow mutable data type

s = {10,20,"Hello",1+3j,20.3}
s
```

```python
## dict

# Key must be immutable

# values may be mutable

# you can access by using the key values

d = {2:45,3:56,4:23}

print(d)

print(d[2])

print(d[3])
```

```python
d1 = {1:"C",2:"Java",3:"Python"}

print(d1)
```

```python
## dict are mutable in nature

d1[2] = "C++"

print(d1)
```

```python
d1.update({4:"C#",5:"VBA"})

print(d1)
```

```python
# Modify

d1[3] = "Madhu"

print(d1)
```

```python
print(d1.values())

print(d1.keys())

print(d1.items())
```
