```python
class BankAccount:
    interest_rate = 5.0

    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.balance = balance

    @classmethod
    def update_interest_rate(cls, new_rate):
        cls.interest_rate = new_rate

    @classmethod
    def show_interest_rate(cls):
        print(f"Current interest rate: {cls.interest_rate}%")

# Usage:
BankAccount.show_interest_rate()  # Output: 5.0%

# Update interest rate
BankAccount.update_interest_rate(6.5)
BankAccount.show_interest_rate()  # Output: 6.5%


a1 = BankAccount("Madhu",3500)
print(a1.account_holder)
print(a1.balance)

print(a1.update_interest_rate(23))
```

```python
class User:
    def __init__(self):
        self.__pin = None
        self.__balance = 0

    def set_pin(self, pin):
        self.__pin = pin
        print("PIN set successfully.")

    def check_balance(self, pin):
        if self.__pin == pin:
            print(f"Your balance is: {self.__balance}")
        else:
            print("Invalid PIN!")

    def deposit(self, pin, amount):
        if self.__pin == pin:
            self.__balance += amount
            print(f"Deposited {amount}. New balance is {self.__balance}")
        else:
            print("Invalid PIN!")

    def withdraw(self, pin, amount):
        if self.__pin == pin:
            if self.__balance >= amount:
                self.__balance -= amount
                print(f"Withdrawn {amount}. Remaining balance is {self.__balance}")
            else:
                print("Insufficient balance!")
        else:
            print("Invalid PIN!")
```
