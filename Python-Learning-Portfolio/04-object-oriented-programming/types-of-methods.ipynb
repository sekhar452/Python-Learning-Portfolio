```python
# Instance Method :
# Static Method
# Class Method

# Instance Method

class BMWCar:
    def __init__(self,name,cc,color):
        self.name = name
        self.cc = cc
        self.color = color
    def start_engine(self):
        print(self.name,"Engine is starting")

def main():
    c = BMWCar("Benz",1200,"Blue")
    print(c.name)
    print(c.cc)
    print(c.color)
    c.start_engine()

if __name__ == "__main__":
    main()
```

```python
# Static Method

class Benz:
    def __init__(self,name,cc,color):
        self.name = name
        self.cc = cc
        self.color = color
    def start_engine(self):
        print(self.name,"Engine is starting")
    
def main():
    B = Benz("Benz",1500,"RED")
    print(B.color)
    print(B.cc)
    print(B.name)

    B.start_engine() # Benz.start_engine(c)

if __name__ == "__main__":
    main()
```

```python
# Static Method

class Benz:
    def __init__(self,name,cc,color):
        self.name = name
        self.cc = cc
        self.color = color
    def start_engine(self):
        print(self.name,"Engine is starting")
    
    @staticmethod
    def Kms_to_miles(kms):
        print(kms*1.6)

def main():
    B = Benz("Benz",1500,"RED")
    print(B.color)
    print(B.cc)
    print(B.name)

    B.start_engine() # Benz.start_engine(c)
    B.Kms_to_miles(2)
    Benz.Kms_to_miles(4) # Static method can called by using class varible reference or class name

if __name__ == "__main__":
    main()
```

```python
# Static Method

class Benz:
    def __init__(self,name,cc,color):
        self.name = name
        self.cc = cc
        self.color = color
    @classmethod
    def x1(cls):
        return cls("X1",1998,"blue")
    @classmethod
    def Series5(cls):
        return cls("5 Series",2993,"blue")
    @classmethod
    def i8(cls):
        return cls("i8",1499,"Red")
    
    def display(self):
        print(self.name)
        print(self.cc)
        print(self.color)
    

def main():
    c1 = Benz.x1()
    c2 = Benz.Series5()
    c3 = Benz.i8()
    c1.display()
    c2.display()
    c3.display()

    

if __name__ == "__main__":
    main()
```
