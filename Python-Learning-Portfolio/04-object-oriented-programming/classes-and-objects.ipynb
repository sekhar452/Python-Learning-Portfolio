```python
## instance method and variables

class Dog:
    ## creating the constructor class
    def __init__(self,name,age):
        self.name = name
        self.age = age

# creation of objects
D = Dog("Buddy",2) # created the object
print(D)
print(D.name,D.age)

D1 = Dog("Lucky",3)
print(D1.name,D1.age)
```

```python
## Define a class with Instance method

class Dog1:
    def __init__(self,name,age): # Instance variables
        self.name = name
        self.age = age
    
    def bark(self): # instance method
        print(f"{self.name} says bow bow")

d = Dog1("Lucky",23) # creation of object
d.bark()
print(d.age) ## calling via reference value
print(d.name)


```

```python
# car is an class and class has state and behaviour brand, cc, color start(), shif_gear(), accelerate()

class car:
    def __init__(self,cc,color,brand):
        self.cc = cc # Instance variables
        self.brand = brand # Instance variables
        self.color = color # Instance variables


    def start_engine(self): 
        print("starting the engine") ## Instance method
    def shif_gear(self):
        print("shifting the gear") # Instance method
    def accelerate(self):
        print("car is accelerating") # Instance method

c = car(1000,"Auddi","YELLow")
c1 = car(1200,"Benz","Blue")

print(c.cc)
print(c.brand)
print(c.color)
c.start_engine()
c.accelerate()
c.shif_gear()
```

```python
## Footballer

## footballer has name, team, goals

## his behaviours are shooting() passing() running()

# creating footballer class

class FootBaller:
    # defining or creating  instance variables 
    def __init__(self,name,team,goals):
        self.name = name
        self.team = team
        self.goals = goals
    def shooting(self):     # defining instance methods
        print(self.name,"is shooting the goal")
    def passing(self):  # defining instance methods
        print(self.name,"is passing the goals")
    def running(self): # defining instance methods
        print(self.name, "is runnig towards goal")   

F = FootBaller("Cristiano","Juvetas",746)  # instalizing instance variables
print(F.name)
print(F.goals)
print(F.team)

F.shooting() # calling instance methods
F.passing()
F.running()
```

```python
## Footballer

## footballer has name, team, goals

## his behaviours are shooting() passing() running()

# creating footballer class

class FootBaller:
    # defining or creating  instance variables 
    def __init__(self,name,team,goals):
        self.name = name
        self.team = team
        self.goals = goals
    def display(self):
        print(self.name)
        print(self.goals)
        print(self.team)
    def shooting(self):     # defining instance methods
        print(self.name,"is shooting the ball towards goal")
    def passing(self):  # defining instance methods
        print(self.name,"is passing the ball")
    def running(self): # defining instance methods
        print(self.name, "is runnig towards goal")   

F3 = FootBaller("Sunil Chhetri","India",94)  # instalizing instance variables

F3.display()

F3.shooting() # calling instance methods
F3.passing()
F3.running()
```

```python
## Static Varibles

# creating a class Aadhar 

# Aadhar has instance varibales name, age, gender, state, nationlaity

class Aadhar: # creation of class 
    def __init__(self,name,age,gender,state,nationlaity): # creating the instance variables
        self.name = name
        self.age = age
        self.gender = gender
        self.state = state
        self.nationlaity = nationlaity
    def display(self): # creating the instance method
        print(self.name)
        print(self.age)
        print(self.gender)
        print(self.state)
        print(self.nationlaity)

A = Aadhar("Rohit",28,"Male","Kar","Indian")
S = Aadhar("Swathi",20,"Female","A P", "Indian")
R = Aadhar("Rita",15,"Female","A P", "Indian")

A.display()
S.display()
R.display()

```
