```python
# Types of inheritance 

# advantages of inheritance

# code resuability, reduced development time, easy maintainbility

# single Inheritance

class Alpha:
    def fun(self):
        print("This is Alpha class and inside fun()")
class Beta(Alpha):
    pass
b = Beta()
b.fun()
```

```python
class Alpha(object):
    def fun(self):
        print("Alpha class fun()")
print(dir(Alpha)) # all these are inherited from the object class
```

```python
# Multi level inheritance:

class Alpha:
    def fun1(self):
        print("Inside Alpha fun1()")
class Beta(Alpha):
    def fun2(self):
        print("Inside beta fun2()")
class Gamma(Beta):
    pass

G = Gamma()
G.fun2()
G.fun1()
print(dir(Gamma))
```

```python
# Multiple inheritance

class Alpha:
    def fun1(self):
        print("alpha class fun1()")

class Beta:
    def fun2(self):
        print("Beta class fun2()")
class Gamma(Alpha,Beta):
    pass

G = Gamma()

G.fun1()
G.fun2()

```

```python
# Multiple inheritance # Here we get amibiguity problem Internally python uses the alogorithm class 

# Method resolution order
# c3 linearlization algorithm 


class Alpha:
    def fun1(self):
        print("alpha class fun1()")

class Beta:
    def fun1(self):
        print("Beta class fun1()")
class Gamma(Alpha,Beta):
    pass

G = Gamma()

G.fun1()

```

```python
# Types of methods

# Inherited Method used for code reusability

# overriden method method customization

# specialized method adding new features

class Messenger:
    def send_message(self):
        print("Text msg is sent")
    def receive_message(self):
        print("Text msg is received")
    
class FacebookMessenger(Messenger): ## Inherited Method
    pass
class WhatsAppMessenger(Messenger):
    def send_message(self):
        print("text, Photos vedious files is sent") # overriden methods
    def receive_message(self):
        print("text, Photo,vedious, files is sent") # overriden methods
    
    def set_dp(self):
        print("dp is sent") # specialized methods
    
    def set_status(self):
        print("status is set")

Fb = FacebookMessenger()
Fb.receive_message()
Fb.send_message()

WM = WhatsAppMessenger()


WM.send_message()
WM.receive_message()
WM.set_dp()
WM.set_status()
```

The super keyword refers to superclass (parent) objects. It is used 
to call superclass methods, and to access the superclass constructor. 
The most common use of the super keyword is to eliminate the 
confusion between super-classes and sub-classes that have methods 
with the same name.

```python
class Customer:
    def __init__(self,name,phone,email):
        self.name = name
        self.phone = phone
        self.email = email
class PlatinumCustomer(Customer):
    def __init__(self, name, phone, email,plat_id):
        self.name = name
        self.phone = phone
        self.email = email
        self.plat_id = plat_id
    def display(self):
        print(self.__dict__)

def main():
    p = PlatinumCustomer("Madhu",9398436534,"madhupranitha96@gmail.com",143143)
    p.display()

if __name__ == "__main__":
    main()
```

In the above example we can reduce the size of __init__() of child 
class i.e. PlatinumCustomer() by using super(). Let us see below how 
to do it

```python
class Customer:
    def __init__(self,name,phone,email):
        self.name = name
        self.phone = phone
        self.email = email
class PlatinumCustomer(Customer):
    def __init__(self, name, phone, email,plat_id):
        super().__init__(name,phone,email)
        self.plat_id = plat_id
    def display(self):
        print(self.__dict__)

def main():
    p = PlatinumCustomer("Madhu",9398436534,"madhupranitha96@gmail.com",143143)
    p.display()

if __name__ == "__main__":
    main()
```

```python
class Customer:
    def __init__(self,name,phone,email):
        self.name = name
        self.phone = phone
        self.email = email
    def place_order(self,dish):
        cost = 0
        del_charge=50
        if dish == "Pizza":
            cost = 500 + del_charge
        else:
            cost = 250 + del_charge
        return cost
class PlatinumCustomer(Customer):
    def __init__(self, name, phone, email,plat_id):
        super().__init__(name,phone,email)
        self.plat_id = plat_id
    def place_order(self,dish):
        del_charge = 50
        return (super().place_order(dish)-del_charge) * 0.95
    
def main():
    p = PlatinumCustomer("Ramya",9393939393,"shaikraja@gmail.com",143143)
    print(p.place_order("Pizza"))

if __name__ == "__main__":
    main()
```
