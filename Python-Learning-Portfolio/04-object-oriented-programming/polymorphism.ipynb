```python
# Advantages of using the polymorphism 
# 1. flexible
# 2.reduce code size

# polymorphic version

class messenger:
    def use_keyboard(self):
        print("using Keyboard")
    def send_message(self):
        print("text message sent")
    def receive_messaage(self):
        print("text message receive")
    
class WhatsApp:
    def use_keyboard(self):
        print("using Keyboard by whats app")
    def send_message(self):
        print("text message sent by whats app")
    def receive_messaage(self):
        print("text message receive by whats app")
    def send_live_location(self):
        print("text messaged via live_location")

class Facebook:
    def use_keyboard(self):
        print("using Keyboard by Facebook")
    def send_message(self):
        print("text message sent  by Facebook")
    def receive_messaage(self):
        print("text message receive by Facebook")
    def use_buitin_apps(self):
        print("text message receive builtin apps")   

class Insta:
    def use_keyboard(self):
        print("using Keyboard by Insta")
    def send_message(self):
        print("text message sent  by Insta")
    def receive_messaage(self):
        print("text message receive by Insta")
    def filter(self):
        print("text message filters")

def use_messenger(ref):
    ref.use_keyboard()
    ref.send_message()
    ref.receive_messaage()


    if type(ref) == WhatsApp:
        ref.send_live_location()
    if type(ref) == Insta:
        ref.filter()
    if type(ref) == Facebook:
        ref.use_buitin_apps()

wa = WhatsApp()
Fb = Facebook()
In = Insta()

use_messenger(wa)
use_messenger(Fb)
use_messenger(In)


```
