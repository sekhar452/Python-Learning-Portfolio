### Types of variables

Global value: The value which is outside the fucntion called the global value

Local Value: The value which is inside the fucntion called the Local value



```python
x = 999

def fun():
    y = 99
    print(y)

print(x)
fun()
```

```python
x = 999

def fun():
    y = 99
    print(y)
    print(x)

fun()
```

```python
## Local scope
x = 990

def fun():
    y = 98
    print(y)
    print(x)

fun()
print(x)
print(y)
```

```python
## Global varibles are stored in gloabls() in the form of dict

x = 78

def fun():
    y = 65
    print(y)
    print(globals())
    print(locals())
fun()
```

```python
# Can i have local varibales and global values with same names

x = 34

def fun():
    y = 34
    print(x)
    print(y)

print(x)
fun()
```

```python
## can i make local varibales as global -- Yes you can by using the global key words 

x = 99

def fun():
    global x
    x = 9999
    print(x)

print(x)
fun()
print(x)
```

```python
## Recursive function

def fact(n):
    if n==1:
        return 1
    else:
        return n*fact(n-1)
print(fact(5))
```
